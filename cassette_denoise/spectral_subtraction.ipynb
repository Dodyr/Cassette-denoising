```python
num_samples = 32
```

```python
for i in tqdm(range(1, num_samples+1)):
    input_path = os.path.join("data/noisy", f"{i}.wav")
    audio, sr = librosa.load(input_path, sr=None, mono=False)
    
    left_channel = audio[0]
    right_channel = audio[1]

    denoised_left = apply_spectral_sub(left_channel, nfft=2048, db_reduc=8, lookback=5)
    denoised_right = apply_spectral_sub(right_channel, nfft=2048, db_reduc=8, lookback=5)

    denoised_audio = np.vstack([denoised_left, denoised_right]).T
    sf.write(f"data/spectral_sub/{i}.wav", denoised_audio, sr)
```
